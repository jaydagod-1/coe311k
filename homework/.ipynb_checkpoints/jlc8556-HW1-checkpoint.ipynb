```python
import random
import numpy as np

def monte_carlo_Pi(num_of_shots): #Defining the orgeration under a function so I can input the value n later through 1000 iterations
    num_of_hits=0 # Allocating a value to hits
    for i in range(0,num_of_shots):
        a=random.uniform(-1,1)
        b=random.uniform(-1,1) #Calling 2 random values between -1 and 1 for i when in range between 0 and the new allocation for shot numbers
        
        if (((a*a)+(b*b))<(1)):
            num_of_hits=num_of_hits+1 #computing a^2+b^2 being less than 1, then increasing the allocation of hits by one as this means it is inside of the circle
            
    return 4*num_of_hits/num_of_shots # Multiply the proportion of hits to shots by 4 and return the value
```

```python
import numpy as np

pi_actual=np.pi

abs_errors=[] 
relative_errors=[]
sample_sizes=[] 
#Consulted Gemini on the best way to store x values, and realized I could do it by using an empty list and then reallocating values back into it later

for n in range(10, 1001, 20):
    calc_pi=monte_carlo_Pi(n) #Plugging in n into the function from earlier
  
    error_actual=abs(pi_actual-calc_pi)
    error_relative=error_actual/pi_actual 
    
    abs_errors.append(error_actual)
    relative_errors.append(error_relative)
    sample_sizes.append(n)
    #Important values needed in order to plot the chart later on
```

```python
plt.figure(figsize=(10, 6))

plt.plot(sample_sizes, abs_errors, 'o-', color='orange') #Consulted Gemini on how to make the plot orange

plt.title('Monte Carlo Error Convergence')
plt.xlabel('Number of Samples (n)')
plt.ylabel('Absolute Error') #Added labels to ensure maximum detail

plt.show()
```
