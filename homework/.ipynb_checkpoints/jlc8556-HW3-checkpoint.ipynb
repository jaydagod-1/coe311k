```python
import numpy as np

sum = 0
pi_actual = np.pi

for n in range(5000):
    denominator = 2*n+1
    
    term = ((-1)**n) / denominator
    
    sum += term
    
    if (n+1) % 100 == 0:
        pi_est = sum*4
        rel_error = abs(pi_actual - pi_est)/pi_actual
        print(f"Terms: {n+1} ; Calculated Pi: {pi_est:.10f} ; Relative Error: {rel_error:}")
```

This is the Generalized Leibniz Series for 𝜋 in Python

```python
import numpy as np
import math as mth
import matplotlib.pyplot as plt

running_sum = 0
x=np.arange(0,10,.1) #Had to use .1 as I can not divide by 0 or else an error pops up. Also had to attribute x to an arange as the power operator can't do a range of values
sin_actual = np.sin(x)
num_of_terms = int(input("Enter the number of terms to calculate: ")) #Took an input here so any number of terms can be calculated. Had to define this as an int as it automatically defaults to a string

for n in range(num_of_terms):
    denom = mth.factorial((2*n)+1)
    numer = ((-1)**n) * (x**(2*n+1)) #divided it into denominator and numerator terms to make it easier to write, calculate, and see errors.
    term = numer/denom
    running_sum += term
    
    sin_est = running_sum
    abs_error = abs(sin_actual-sin_est)
    
    print(f"Term: {n+1} ; Calculated Sin: {sin_est:} ; Absolute Error: {abs_error:}") #Couldnt clean up printed values due to np.arange not being compatable with that
    
    plt.plot(x,abs_error,label=f'Number of Terms Calculated: {n+1}') # Originally, I placed this outside the loop, but realized it could be placed INSIDE the loop so it adds a new line 4 different times
    
plt.yscale('log')
plt.legend()
plt.title("The Absolute Error Sin(x) Calculated by the Taylor Series as Terms Increase")
plt.xlabel("Value of X")
plt.ylabel("Absolute Error")
plt.show()
```

Discuss how the truncation error behaves as you include more terms. Does the error decrease uniformly for all values of x? Does the truncation error behave differently for small vs. large values of x (try feeding larger values of x)? (a few sentences is fine)

For just the first term, the absolute error stabilizes and doesn't increase much as the value of x increases. However, as the number of terms increase, the size of the error grows as x increases, and that rate of how much the error grows increases as the number of terms increase as well.
