```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
k = 1.0  # Constant in the equation (e.g., spring constant)
h_values = [0.01, 0.02, 0.03, 0.05, 0.1, 0.2, 0.3, 0.5]  # Trialing Step size
t_start = 0  # Start of the interval
t_end = 20  # End of the interval
y_0 = 1.0  # Initial condition for y
v_0 = 0.0  # Initial condition for dy/dx

def euler_method(h, k, t_start, t_end, y_0, v_0):
    t_values = np.arange(t_start, t_end + h, h)
    y_vals = np.zeros(len(t_values))
    v_vals = np.zeros(len(t_values))
    
    y_vals[0] = y_0
    v_vals[0] = v_0
    
    for i in range(1, len(t_values)):
        y_vals[i] = y_vals[i-1] + h * v_vals[i-1]
        v_vals[i] = v_vals[i-1] + h * (-k * y_vals[i-1])  
    return t_values, y_vals
```

```python
# Plotting everything on one canvas
plt.figure(figsize=(12, 8))

for h in h_values:
    t, y = euler_method(h, k, t_start, t_end, y_0, v_0)
    plt.plot(t, y, label=f'h={h}')

# Plotting the "True" solution (cos(sqrt(k)*t)) for reference
t_exact = np.linspace(t_start, t_end, 500)
y_exact = np.cos(np.sqrt(k) * t_exact)
plt.plot(t_exact, y_exact, 'k', alpha=0.5, lw=2, label='Exact (Analytical)')

plt.xlabel('Time (t)')
plt.ylabel('Displacement (y)')
plt.title(f'Euler Stability Comparison (k={k})')
plt.legend(loc='upper right', bbox_to_anchor=(1.15, 1))
plt.grid(True)
plt.show()
```
