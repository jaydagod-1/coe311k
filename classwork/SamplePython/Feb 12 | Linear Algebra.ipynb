Linear Algebra in Python

```python
import numpy as np
```

```python
v1=np.array([1,2,3])# A 3D Vector
v2=np.array([4,5,6])# Another 3D Vector
```

```python
print(v1)
print(v2)
```

```python
v_add=v1+v2
v_sub=v1-v2
```

```python
print(v_add)
print(v_sub)
```

```python
scale = 2
v_scale = scale*v1
```

```python
print(v_scale)
```

```python
dot_prod=np.dot(v1,v2)
```

```python
print(dot_prod)
```

```python
mag=np.linalg.norm(v1)
```

```python
print("Magnitude of V1:", mag)
```

```python
A=np.array([[1,2,3],[4,5,6]])
print(A)
print()

B=np.array([[7,8,9],[10,11,12],[13,14,15]])
print(B)
```

```python
C=np.array([[1,2,3],[4,5,6]])
matrix_add=A+C
matrix_sub=A-C

print("A+C=\n", matrix_add)
print("A-C=\n", matrix_sub)
```

```python
matrix_mul=np.dot(A,B)
print(matrix_mul)
```

```python
transpose_A = A.T
print("Transpose of A:\n", transpose_A)
```

```python
a=np.array([[2,3],[4,-1]])
b=np.array([5,3])
print(a)
print()
print(b)
x=np.linalg.solve(a,b)
print(x)
```

```python
A=np.array([[2,1],[1,2]])
e_val,e_vec=np.linalg.eig(A) # 2 values because it returns 2 values
```

```python
print("Eval:",e_val)
print("Evec:",e_vec)
```

```python
A = np.array([[4, 7], [2, 6]])

A_inv = np.linalg.inv(A)
print("Inverse of A:\n", A_inv)

identity = np.dot(A, A_inv)
print("A * A_inv:\n", identity)
```
