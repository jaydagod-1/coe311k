```python
import numpy as np
import matplotlib.pyplot as plt

a, b = 0, 2 * np.pi
N = 50
x = np.linspace(a, b, N)
h = (b - a) / (N - 1)

f = lambda x: np.sin(x)
df_true = np.cos(x)
```

## Central Difference Approximation

```python
df_central = (f(x + h) - f(x - h)) / (2 * h)
```

## Forward Difference Approximation

```python
df_forward = (f(x + h) - f(x)) / h
```

## Backward Difference Approximation

```python
df_backward = (f(x) - f(x - h)) / h
```

## Plotting

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 1, 1)
plt.plot(x, df_true, 'k-', lw=3, label='True Derivative (cos(x))', alpha=0.3)
plt.plot(x, df_central, 'g.', label='Central Difference')
plt.plot(x, df_forward, 'r--', label='Forward Difference')
plt.plot(x, df_backward, 'b--', label='Backward Difference')
plt.title('Comparison of Finite Difference Methods')
plt.legend()
plt.grid()

plt.tight_layout()
plt.show()
```
