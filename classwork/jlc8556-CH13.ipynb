```python
import numpy as np
import time
import matplotlib.pyplot as plt

def naive_matrix_multiplication(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]
    return C
```

```python
sizes = []
naive_times = []
numpy_times = []
current_size = 3


for i in range(8):
    A = np.random.rand(current_size, current_size)
    B = np.random.rand(current_size, current_size)
    sizes.append(current_size)
    
    start = time.perf_counter()
    naive_matrix_multiplication(A, B) 
    end = time.perf_counter()
    naive_times.append(end - start)

    start = time.perf_counter()
    np.dot(A, B)
    end = time.perf_counter()
    numpy_times.append(end - start)

    current_size *= 2
```

```python
plt.figure(figsize=(10, 6))
matrix_numbers = range(1, 9)

plt.plot(matrix_numbers, naive_times, 'ko-', label='Custom Naive (O(n^3))')
plt.plot(matrix_numbers, numpy_times, 'cs-', label='Optimized NumPy')

plt.title('Matrix Multiplication: Custom vs. NumPy Runtime')
plt.xlabel('Matrix Pair Number (1 to 8)')
plt.ylabel('Time in Seconds')
plt.legend()
plt.grid(True)
plt.show()
```
