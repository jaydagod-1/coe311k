```python
import numpy as np

def gauss_seidel(A, b, x0, tol, max_its):
    n = len(b)
    x = x0.copy().astype(float)
    
    beginning_time = time.perf_counter()
    
    for k in range(max_its):
        x_new = x.copy()
        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i)) # Using already updated values
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n)) # Using old values
            x_new[i] = (b[i] - s1 - s2) / A[i][i]
        
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            runtime=time.perf_counter()-beginning_time
            return x_new,k+1,runtime
        
        x = x_new
        
    return x, max_its, time.perf_counter()-beginning_time

A = np.array([[1,2,3,0,0],
              [2,1,2,3,0],
              [3,2,1,2,3],
              [0,3,2,1,2],
             [0,0,3,2,1]], dtype=float)

b=np.array([14,22,33,26,22])

x0 = np.zeros_like(b)
tol = 1e-6
max_its = 100
solution, iterations, duration = gauss_seidel(A, b, x0, tol, max_its)
print(f"Solution: {solution}")
print(f"Iterations: {iterations}")
print(f"Time Taken: {duration:.8f} seconds")
```
