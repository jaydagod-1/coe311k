```python
import numpy as np

def setup_tridiagonal_matrix(x, y):
    n = len(x)
    h = np.diff(x)  # Step sizes between data points
    
    size = n - 2
    A = np.zeros(size)  # Subdiagonal
    B = np.zeros(size)  # Main diagonal
    C = np.zeros(size)  # Superdiagonal
    D = np.zeros(size)  # Right-hand side vector

    # Populate the tridiagonal matrix coefficients
    for i in range(1, n - 1):
        idx = i - 1  
        if idx > 0:
            A[idx] = h[i-1]
            B[idx] = 2 * (h[i-1] + h[i])
        if idx < size - 1:
            C[idx] = h[i]
        term1 = (y[i+1] - y[i]) / h[i]
        term2 = (y[i] - y[i-1]) / h[i-1]
        D[idx] = 6 * (term1 - term2)
    return A, B, C, D

x = np.array([0, 1, 2, 3, 4, 5])
y = np.array([0, 1, 0, 1, 0, 1])

A, B, C, D = setup_tridiagonal_matrix(x, y)

print("Subdiagonal (A):", A)
print("Main diagonal (B):", B)
print("Superdiagonal (C):", C)
print("Right-hand side (D):", D)
```
