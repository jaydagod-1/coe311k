```python
def diagonally_dominance(A):
    diag = np.abs(np.diag(A))
    off_diag = np.sum(np.abs(A), axis=1) - diag
    return np.all(diag >= off_diag)

A1 = np.array([[0.582745, 0.48, 0.10, 0, 0], 
                  [0.48, 1.044129, 0.46, 0.10, 0], 
                  [0.10, 0.46, 1.10431, 0.44, 0.10], 
                  [0, 0.10, 0.44, 0.963889, 0.42], 
                  [0, 0, 0.10, 0.42, 0.522565]])

A2 = np.array([[1, 1], [1, 1.0001]])

for mat in [A1, A2]:
    print(f"Is the matrix diagonally dominant? {diagonally_dominance(mat)}")
    print(mat)
```
