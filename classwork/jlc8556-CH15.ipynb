```python
import numpy as np
import matplotlib.pyplot as plt

# Function representing dy/dx = -2y
def f(x, y):
    return -2 * y

x_start, x_end = 0, 2
first_run_x, first_run_y = None, None
plt.figure(figsize=(12, 6))

h = 0.01
while h < 1.01:
    num_steps = int((x_end - x_start) / h)
    x_values = np.linspace(x_start, x_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)
    y_values[0] = 1 # Initial condition
    
    # Implement Euler's method
    for i in range(num_steps):
        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])

    if first_run_x is None:
        first_run_x = x_values
        first_run_y = y_values

    y_reference  = np.interp(x_values, first_run_x, first_run_y)
    point_errors = np.abs(y_values - y_reference)
    total_error  = np.sum(point_errors)

    #Prints the error(current run vs first run) for each point, and the total accumulated error for each run.
    print(f"h={h:.2f} ; Point errors: {np.round(point_errors, 4)} ; Total error: {total_error:.5f}")
    plt.plot(x_values, y_values, alpha=0.6, label=f"h={h:.2f}") 

    h += 0.05
    
plt.xlabel('x')
plt.ylabel('y')
plt.title("Approximation of dy/dx = -2y using Euler's Method")
plt.legend()
plt.show()
```
