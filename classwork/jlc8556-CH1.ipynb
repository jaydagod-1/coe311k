```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(-10,10,100)
b = np.linspace(-5,5,100)
c = x**3
a = np.linspace(-1,1,100)
d = np.linspace(.001,10,100)

plt.subplot(2, 2, 1) # (row, column, active area)
plt.plot(b, np.sin(3*b), 'r^')
plt.title("Sine Wave")

plt.subplot(2, 2, 2)
randdata = np.random.randn(500)
plt.hist(randdata, bins=30, color='y', edgecolor='k')
plt.title("Bell Curve")

plt.subplot(2,2,3)
plt.scatter(b,c,s=5, color='c')
plt.title("Bell Curve")

plt.subplot(2,2,4)
plt.plot(x, np.log(d), 'm')
plt.title("Logarithmic Curve")

plt.subplots_adjust(hspace=.6)
plt.show()
```
