```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import CubicSpline

x = np.array([0, 1, 2, 3, 4, 5])
y = np.array([0, 1, 0, 1, 0, 1])

cs = CubicSpline(x, y, bc_type='natural')

poly_coeffs = np.polyfit(x, y, 5)
poly_func = np.poly1d(poly_coeffs)

x_interp = np.linspace(0, 5, 200)
y_interp = cs(x_interp)
y_poly = poly_func(x_interp)
```

```python
plt.plot(x, y, 'ko', label="Original Data Points", markersize=8)
plt.plot(x_interp, y_interp, 'k-', linewidth=2, label="Natural Cubic Spline")
plt.plot(x_interp, y_poly, 'c--', linewidth=2, label="5th-Degree Polynomial")

plt.title("Cubic Spline vs. Polynomial Fitting")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```
