```python
import random

def montecarloPi(numofshots):
    numofhits=0
    for i in range(0,numofshots):
        a=random.uniform(-1,1)
        b=random.unifrom(-1,1)
        if (((a*a)+(b*b))<(1)):
            numofhits=numofhits+1
            
    return 4*numofhits/numofshots
```

```python
print(montecarloPi)
```
