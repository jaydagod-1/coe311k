```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(-10,10,100)
y = np.sin(x)

plt.subplot(2, 1, 1) # (row, column, active area)
plt.plot(x, np.sin(3*x), 'r-')

plt.subplot(2, 2, 2)
plt.hist(x, 50)

plt.subplot(2,2,3)
y = np.cos(x)
plt.scatter(x,y)

plt.subplot(2,2,4)
plt.plot(x, np.arccos(x), 'c')

plt.show()
```
