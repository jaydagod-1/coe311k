```python
import random
import numpy as np

def montecarloPi(numofshots):
    numofhits=0
    for i in range(0,numofshots):
        a=random.uniform(-1,1)
        b=random.uniform(-1,1)
        if (((a*a)+(b*b))<(1)):
            numofhits=numofhits+1
            
    return 4*numofhits/numofshots
```

```python
import numpy as np

piactual=np.pi

abserrors = []
relerrors = []

for i in range(100):
    calcpi = montecarloPi(1000)
  
    erroractual=abs(piactual-calcpi)
    errorrelative=erroractual/piactual
    
    abserrors.append(erroractual)
    relerrors.append(errorrelative)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))


plt.subplot(1, 2, 1)
plt.plot(abserrors,'b')
plt.title('Absolute Error')
plt.xlabel('Runs')


plt.subplot(1, 2, 2)
plt.plot(relerrors,'r')
plt.title('Relative Error')
plt.xlabel('Runs')

plt.tight_layout()
plt.show()
```
