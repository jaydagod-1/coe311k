```python
import numpy as np
import matplotlib.pyplot as plt

y0 = 0
t0 = 0
tf = 30
h_sizes = [0.1, 0.01, 0.001, .0000001]

def f(t, y):
    return -1000 * y + 3000 - 2000 * np.exp(-t)

def analytical_sol(t):
    return 3 - 0.998 * np.exp(-1000 * t) - 2.002 * np.exp(-t)

def forward_euler_safe(f, y0, t0, tf, h): #This method helps prevent the overflow issues
    num_steps = int((tf - t0) / h)
  
    save_every = max(1, num_steps // 1000)
    
    t_vals_out = []
    y_vals_out = []
    
    t = t0
    y = y0
    
    for i in range(num_steps + 1):
        if i % save_every == 0:
            t_vals_out.append(t)
            y_vals_out.append(y)
        
        y_next = y + h * f(t, y)
        t = t + h
        y = y_next
        
        #Will stop if it explodes
        if not np.isfinite(y) or abs(y) > 1e5:
            break
            
    return np.array(t_vals_out), np.array(y_vals_out)
```

```python
plt.figure(figsize=(10, 6))

# Analytical Plot
t_fine = np.linspace(t0, tf, 1000)
plt.plot(t_fine, analytical_sol(t_fine), 'k-', label='Analytical', linewidth=3)

for h in h_sizes:
    t_vals, y_vals = forward_euler_safe(f, y0, t0, tf, h)
    if len(y_vals) > 1:
        plt.plot(t_vals, y_vals, label=f'Euler h={h}')

plt.title("Forward Euler vs Analytical Solution (Stiff ODE)")
plt.xlabel("t")
plt.ylabel("y(t)")
plt.ylim(-1, 5) # Crucial to see the solution when h=0.1 blows up
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
