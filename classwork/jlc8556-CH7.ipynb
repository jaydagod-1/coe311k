```python
import numpy as np

def jacobi(A, b, x0, tol=1e-6, max_its=1000):
    n = len(b)
    x = x0.copy().astype(float)
    for k in range(max_its):
        x_new = np.zeros_like(x)
        for i in range(n):
            s = sum(A[i][j] * x[j] for j in range(n) if j != i)
            x_new[i] = (b[i] - s) / A[i][i]
        
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k + 1
        x = x_new
    return x, max_its
```

```python
def gauss_seidel(A, b, x0, tol=1e-6, max_its=1000):
    n = len(b)
    x = x0.copy().astype(float)
    for k in range(max_its):
        x_old = x.copy()
        for i in range(n):
            s1 = sum(A[i][j] * x[j] for j in range(i))
            s2 = sum(A[i][j] * x_old[j] for j in range(i + 1, n))
            x[i] = (b[i] - s1 - s2) / A[i][i]
        
        if np.linalg.norm(x - x_old, ord=np.inf) < tol:
            return x, k + 1
    return x, max_its
```

```python

A1 = np.array([[0.582745, 0.48, 0.10, 0., 0.],
               [0.48, 1.044129, 0.46, 0.10, 0.],
               [0.10, 0.46, 1.10431, 0.44, 0.10],
               [0., 0.10, 0.44, 0.963889, 0.42],
               [0., 0., 0.10, 0.42, 0.522565]], dtype=float)
b1 = np.array([1.162745, 2.084129, 2.20431, 1.923889, 1.042565], dtype=float)

A2 = np.array([[1, 1], [1, 1.0001]], dtype=float)
b2 = np.array([1, 1], dtype=float)

A3 = A2.copy()
b3 = np.array([1, 1.0001], dtype=float)

systems = [("System 1", A1, b1), ("System 2", A2, b2), ("System 3", A3, b3)]

for name, A, b in systems:
    x0 = np.zeros(len(b))
    
    sol_j, it_j = jacobi(A, b, x0)
    print(f"Jacobi:       {it_j:4} iterations ; Sol: {sol_j}")
    
    sol_gs, it_gs = gauss_seidel(A, b, x0)
    print(f"Gauss-Seidel: {it_gs:4} iterations ; Sol: {sol_gs}")
```
