Part A

```python
import numpy as np

X1 = np.array([1, 2, 3, 4, 5])
Y1 = np.array([1.2, 2.8, 3.6, 4.5, 5.1])


A1 = np.vstack([X1, np.ones(len(X1))]).T
m1, c1 = np.linalg.lstsq(A1, Y1, rcond=None)[0]

print(f"Slope: {m1}, Intercept: {c1}")
```

```python
import matplotlib.pyplot as plt

plt.scatter(X1, Y1, color='blue', label='Data points')
plt.plot(X1, m1*X1 + c1, color='red')


plt.title('Best Fit line')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```

Part B

```python
X_rand = np.random.permutation(X1)
Y_rand = np.random.permutation(Y1)

A_rand = np.vstack([X_rand, np.ones(len(X_rand))]).T
m_rand, c_rand = np.linalg.lstsq(A_rand, Y_rand, rcond=None)[0]

plt.scatter(X_rand, Y_rand, color='teal', label='Randomized Data')
plt.plot(X_rand, m_rand*X_rand + c_rand, color='orange', label='Random Fit')
plt.title('Randomized Best Fit Line')
plt.legend()
plt.show()
```

Part C

```python
X_synth = np.linspace(0, 1, 101)
noise = np.random.normal(0, 0.2, len(X_synth))
Y_synth = 3.0 * X_synth + 2.0 + noise

A_synth = np.vstack([X_synth, np.ones(len(X_synth))]).T
m_synth, c_synth = np.linalg.lstsq(A_synth, Y_synth, rcond=None)[0]
```

```python
plt.scatter(X_synth, Y_synth, color='yellow', alpha=0.5, label='Synthetic Noisy Data')
plt.plot(X_synth, m_synth*X_synth + c_synth, color='green', label='Recovered Line')
plt.title('Part 3: Custom Equation + Noise')
plt.legend()
plt.show()
```
